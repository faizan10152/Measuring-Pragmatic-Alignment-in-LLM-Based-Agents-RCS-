```python
import pandas as pd

df = pd.read_excel("/content/Annotated_Data.xlsx")
```

```python
from sklearn.preprocessing import LabelEncoder
import joblib

LABEL_COLS = ["STANCE", "ACTION", "PERSONALNESS", "POLITENESS"]

encoders = {}

for col in LABEL_COLS:
    le = LabelEncoder()
    le.fit(df[col].astype(str))   # IMPORTANT: same column as training
    encoders[col] = le

    print(f"{col} classes:", list(le.classes_))
```

```python
joblib.dump(encoders, "/content/label_encoders.pkl")
```
