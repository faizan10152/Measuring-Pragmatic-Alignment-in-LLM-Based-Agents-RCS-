Fits a `LabelEncoder` for each pragmatic dimension on the annotated dataset and serialises the encoders to `notebooks/label_encoders.pkl`.  
These encoders are required by both the critic-training notebook and the agentic pipeline.

> **Dependency:** Run notebook 2 first, or make sure `annotated_data.xlsx` is in `data/`.

```python
import pandas as pd

df = pd.read_excel("../data/annotated_data.xlsx")
```

```python
from sklearn.preprocessing import LabelEncoder
import joblib

LABEL_COLS = ["STANCE", "ACTION", "PERSONALNESS", "POLITENESS"]

encoders = {}

for col in LABEL_COLS:
    le = LabelEncoder()
    le.fit(df[col].astype(str))   # IMPORTANT: same column as training
    encoders[col] = le

    print(f"{col} classes:", list(le.classes_))
```

```python
joblib.dump(encoders, "../notebooks/label_encoders.pkl")
```
